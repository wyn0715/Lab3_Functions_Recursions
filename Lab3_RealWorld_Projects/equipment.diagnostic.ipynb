```python
# Equipment Diagnostic System

LAST_NAME = "WALES"
FAVORITE_ARTIST = "IVOS"
STUDENT_ID = "TUPM-26-7925"

SEED_NUM = int(STUDENT_ID[-1])

print("Student:", LAST_NAME)
print("Favorite Artist:", FAVORITE_ARTIST)
print("Seed Number:", SEED_NUM)
```

```python
# Generate Equipment Readings

def generate_readings(last_name, seed_num, artist):
    voltage = 220 + seed_num
    current = len(last_name) + seed_num
    temperature = 25 + len(artist)

    return { 
        "Voltage": voltage,
        "Current": current, 
        "Temperature": temperature
    }

readings = generate_readings(
    LAST_NAME,
    SEED_NUM,
    FAVORITE_ARTIST
)

print("Generated Equipment Data")
print(readings)
```

```python

# Validate Equipment Readings

def validate_readings(readings):
    try:
        if not 200 <= readings["Voltage"] <= 240:
            raise ValueError("Voltage is out of range")

        if not 0 < readings["Current"]:
            raise ValueError("Current is out of range")

        if not 0 <= readings["Temperature"]:
            raise ValueError("Temperature is out of range")

        return "VALID"

    except (KeyError, TypeError, ValueError) as error:
        return f"INVALID: {error}"


validation_result = validate_readings(readings)

print("Validation Result:", validation_result)
```

```python
# Calculate Equipment Values

def calculate_readings(readings):
    power = readings["Voltage"] * readings["Current"]
    temperature_f = (readings["Temperature"] * 9/5) + 32

    return {
        "Power": power,
        "Temperature_F": temperature_f
    }


calculated_data = calculate_readings(readings)

print("Calculated Data:")
print(calculated_data)
```

```python
# Classify Equipment Condition

def classify_equipment(validation, calculated_data):
    if validation != "VALID":
        return "FAULT"

    power = calculated_data["Power"]
    temperature_f = calculated_data["Temperature_F"]

    if power <= 2500 and temperature_f <= 100:
        return "NORMAL"
    elif power <= 3000 and temperature_f <= 140:
        return "WARNING"
    else:
        return "CRITICAL"

equipment_condition = classify_equipment(
   validation_result,
   calculated_data
)

print("Equipment Condition:" ,equipment_condition)
```

```python
# Diagnostic Logger Decorator

def diagnostic_logger(func):
    def wrapper(*args, **kwargs):
        print("Strarting diagnostic process...:")
        result = func(*args, **kwargs)
        print("Diagnostic process complete.")
        return result

    return wrapper
```

```python
student = "WALES"
favorite_artist = "IVOS"
seed_number = 5
```

```python
# Run Complete Diagnostic

@diagnostic_logger
def run_diagnostic(readings):
    validation = validate_readings(readings)

    if validation != "VALID":
        return {
            "Validation": validation,
            "Condition": "FAULT"
        }

    calculated = calculate_readings(readings)
    condition = classify_equipment(validation, calculated)

    return {
        "Validation": validation,
        "Calculated": calculated,
        "Condition": condition
    }

diagnostic_result = run_diagnostic(readings)

print("Diagnostic Result:")
print(diagnostic_result)
```

```python
# Diagnostic Summary

print("=" * 45)
print("EQUIPMENT DIAGNOSTIC SUMMARY")
print("=" * 45)

print("Student:", student)
print("Favorite Artist:", favorite_artist)
print("Seed Number:", seed_number)

print("\nEquipment Readings:")
for key, value in readings.items():
    print(f"{key}: {value}")

print("\nDiagnostic Results:")
print("Validation:", diagnostic_result["Validation"])

if "Calculated" in diagnostic_result:
    print("Calculated:", diagnostic_result["Calculated"])

print("Condition:", diagnostic_result["Condition"])

print("=" * 45)
```
